Hacer un analisis de esta base de datos utilizando las herramientas que hemos apreendido anteriormente [Medidas de tendencia central y Medidas de dispersión] según el tipo de la variable.

Hacer histograma de las variables categoricas nominales.

Elegir por lo menos una variable categorica nominal y calcular las [Medidas de tendencia central y Medidas de dispersión] para las demás variables por los grupos de la variable categorica elegida - mirar el ejemplo del loop for utilizado anteriormente en este código. ---

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://github.com/manoelgadi/PSDMA/raw/refs/heads/main/datasets/census.csv"
df = pd.read_csv(url, skipinitialspace=True)
print(df.shape)
df.head()
```

```python
print(df.dtypes)
```

```python
numericas = ["age", "hours_per_week", "education_num"]
nominales = ["workclass", "marital_status", "occupation", "relationship",
             "race", "gender", "native_country", "greater_than_50k"]
```

```python
resumen = pd.DataFrame({
    "media": df[numericas].mean(),
    "mediana": df[numericas].median(),
    "moda": df[numericas].mode().iloc[0],
    "desv_tipica": df[numericas].std(),
    "varianza": df[numericas].var(),
    "rango": df[numericas].max() - df[numericas].min(),
    "IQR": df[numericas].quantile(0.75) - df[numericas].quantile(0.25),
})
resumen
```

```python

for col in nominales + ["education"]:
    print("Moda:", df[col].mode()[0])
    print("Nº de categorías:", df[col].nunique())
    print(df[col].value_counts(normalize=True).round(3))
    print()
```

```python
for grupo in df["gender"].unique():
    print("===== gender =", grupo, "=====")
    datos = df[df["gender"] == grupo]
    print("Media:")
    print(datos[numericas].mean())
    print("\nMediana:")
    print(datos[numericas].median())
    print("\nDesviación típica:")
    print(datos[numericas].std())
    print("\nProporción que gana >50k:", round(datos["greater_than_50k"].mean(), 3))
    print("Moda de occupation:", datos["occupation"].mode()[0])
    print()
```

```python
df.groupby("gender")[numericas].agg(["mean", "median", "std"])
```

```python

import matplotlib.pyplot as plt

# Histogramas (gráficos de barras) de las variables categóricas nominales
for col in nominales:
    plt.figure(figsize=(10, 4))
    df[col].value_counts().plot(kind="bar", color="steelblue")
    plt.title("Frecuencia de " + col)
    plt.ylabel("Número de personas")
    plt.xticks(rotation=75)
    plt.tight_layout()
    plt.show()
```
