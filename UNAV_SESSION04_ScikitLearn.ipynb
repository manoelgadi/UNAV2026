### Before going any further, let' play with Scikit-Learn once first!

Create a Python code for a logistic regression of sklearn using the variables Pclass, Sex and Embarked from the Titanic dataset directly from an internet source or GitHub, transform the variables into dummy variables. Set the seed so that all variables always get the same value. Split the data into training and testing. Print the classification report of the model.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Load the Titanic dataset
url = 'https://raw.githubusercontent.com/manoelgadi/PSDMA/refs/heads/main/datasets/titanic_data.csv'
titanic_data = pd.read_csv(url)

# Transform categorical variables into dummy variables
titanic_data = pd.get_dummies(titanic_data, columns=['Pclass', 'Sex', 'Embarked'], drop_first=True)

# Define the features and target variable
X = titanic_data[['Pclass_2', 'Pclass_3', 'Sex_male', 'Embarked_Q', 'Embarked_S']]
y = titanic_data['Survived']

# Set the seed for reproducibility
np.random.seed(42)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create and fit the logistic regression model
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Print the classification report
print(classification_report(y_test, y_pred))
```

### a quick helper code for you: a trick to get only columns with certain types

```python
def get_specific_columns(df, data_types, to_ignore = list(), ignore_target = False):
    columns = df.select_dtypes(include=data_types).columns
    if ignore_target:
        columns = filter(lambda x: x not in to_ignore, list(columns))
    return columns
```

### Processing/Cleaning - Standard Scaler

http://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html
- Standardize features by removing the mean and scaling to unit variance

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
"-----------------------"
import pandas as pd
df = pd.read_excel("https://github.com/manoelgadi/PSDMA/raw/refs/heads/main/datasets/yahoo.xlsx")

variables = get_specific_columns(df, ["float64", "int64"], [target], ignore_target = True)
for v in variables:
    print(v, "- before scaling: mean= {:.10f}".format(df[v].values.mean()), " - min: {:.10f}".format(df[v].values.min()), " - max: {:.10f}".format(df[v].values.max()))
    df[v] = scaler.fit_transform(df[[v]])
    print(v, "- after scaling: mean= {:.10f}".format(df[v].values.mean()), " - min: {:.10f}".format(df[v].values.min()), " - max: {:.10f}".format(df[v].values.max()))
    print("-"*50)
```

### Processing - MinMaxScaler

http://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.MinMaxScaler.html#sklearn.preprocessing.MinMaxScaler
- This estimator scales and translates each feature individually such that it is in the given range on the training set, i.e. between zero and one.

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
"-----------------------"
import pandas as pd
df = pd.read_excel("https://github.com/manoelgadi/PSDMA/raw/refs/heads/main/datasets/yahoo.xlsx")

variables = get_specific_columns(df, ["float64", "int64"], [target], ignore_target = True)
for v in variables:
    print(v, "- before scaling: mean= {:.10f}".format(df[v].values.mean()), " - min: {:.10f}".format(df[v].values.min()), " - max: {:.10f}".format(df[v].values.max()))
    df[v] = scaler.fit_transform(df[[v]])
    print(v, "- after scaling: mean= {:.10f}".format(df[v].values.mean()), " - min: {:.10f}".format(df[v].values.min()), " - max: {:.10f}".format(df[v].values.max()))
    print("-"*50)
```

### Hyper Parameter Optimization - Grid Search Fine Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
import numpy as np

grid = {
    "n_estimators": [10, 50],
    # 'auto' is no longer a valid option for max_features in newer sklearn versions
    "max_features": ['sqrt', 'log2'],
    "bootstrap": [True, False],
    "random_state": [42]
}

"Sklearn"
"-----------------------"

# Drop rows with NaN values in the target variable before splitting
df_cleaned = df.dropna(subset=[target])
X_cleaned = df_cleaned[variables]
y_cleaned = df_cleaned[target]

# Split the cleaned data
df_train, df_test = splitter(df_cleaned, test_size = 0.2, random_state = 42)
X_train= df_train[variables]
y_train = df_train[target]
X_test= df_test[variables]
y_test = df_test[target]

grid_search = GridSearchCV(RandomForestRegressor(), param_grid = grid, cv = 3)
grid_search.fit(X_train, y_train)
optimal_model = grid_search.best_estimator_
"-----------------------"

print("Fine Tuned Model: {0}".format(optimal_model))
```
